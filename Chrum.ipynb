```python
df.rename(columns={'name':'customer_name'},inplace=True)
```

```python
state_contry_mapping = df.dropna(subset=['country']).set_index('state')['country'].to_dict()
df['country'] = df['country'].fillna(df['state'].map(state_contry_mapping))
```
